```python
def remove_outliers_iqr(df,exclude_column):
  for col in df.columns:
    if col==exclude_column:
      continue
    q1=df[col].quantile(0.25)
    q3=df[col].quantile(0.75)
    iqr=q3-q1
    lower_bound=q1-1.5*iqr
    upper_bound=q3+1.5*iqr

    df=df[(df[col] >=lower_bound)& (df[col] <=upper_bound)]
  return df

df_cleaned=remove_outliers_iqr(df,exclude_column='gold_price')
df=df_cleaned
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Exclude 'date' from features for scaling, as MinMaxScaler expects numerical data
features=['gold_price', 'crude_oil_price', 'inflation_rate','interest_rate', 'silver_price', 'festival_season', 'market_trend','demand_index']

scaler=MinMaxScaler()

scaled_data=scaler.fit_transform(df[features])

scaled_df=pd.DataFrame(scaled_data,columns=features)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score
import pandas as pd

x=df[['crude_oil_price', 'inflation_rate',
       'interest_rate', 'silver_price', 'festival_season', 'market_trend',
       'demand_index']]
y=df['gold_price']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)

model=LinearRegression()
model.fit(x_train,y_train)

y_pred=model.predict(x_test)
accuracy=model.score(x_test,y_test)
print(f'Baseline model accuracy:{accuracy:.2f}')
```

# **1.LINEAR REGRESSION**

```python
from sklearn.model_selection import KFold

X=df[['crude_oil_price', 'inflation_rate',
       'interest_rate', 'silver_price', 'festival_season', 'market_trend',
       'demand_index']]
y=df['gold_price']

# Track best fold data
best_accuracy = 0
best_fold = -1

# Initialize KFold
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Save best fold's data
best_X_train = None
best_y_train = None
best_X_test = None
best_y_test = None
best_y_pred = None

# Loop over each fold
for fold, (train_index, test_index) in enumerate(kf.split(X), 1):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    # Train model
    model.fit(X_train, y_train)

    # Predict
    y_pred = model.predict(X_test)
    acc = model.score(X_test, y_test)
    print(f"Fold {fold} Accuracy: {acc}")

    # Update best fold
    if acc > best_accuracy:
        best_accuracy = acc
        best_fold = fold
        best_X_train = X_train
        best_y_train = y_train
        best_X_test = X_test
        best_y_test = y_test
        best_y_pred = y_pred

# Output best fold details
print(f"\n Best Fold: {best_fold} with Accuracy: {best_accuracy}")
```

# **3.SVR**

```python
from sklearn.svm import SVR
model = SVR()
model.fit(best_X_train,best_y_train)
model.score(best_X_test,best_y_test)
```

# **4. RANDOM FOREST**

```python
from sklearn.ensemble import RandomForestRegressor
model=RandomForestRegressor()
model.fit(best_X_train,best_y_train)
model.score(best_X_test,best_y_test)
```

# **5. ADABOOST**

```python
from sklearn.ensemble import AdaBoostRegressor
model=AdaBoostRegressor()
model.fit(best_X_train,best_y_train)
model.score(best_X_test,best_y_test)
```

```python
from  xgboost import XGBRegressor
model=XGBRegressor()
model.fit(best_X_train,best_y_train)
model.score(best_X_test,best_y_test)
```
