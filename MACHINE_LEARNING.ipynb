# LINEAR REGRESSION

## 1.SIMPLE LINEAR REGRESSION

Predicts a dependent variable (output) using **one independent variable(input)**

### Example 1

```python
# step 1:- import libraries:-

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# step 2:- create numpy arrays area and price.
area = np.array([1500, 2000, 2500, 3000, 3500]).reshape(-1, 1)  # Independent variable (X)
price = np.array([300000, 400000, 500000, 600000, 700000])
```

```python
# step 3:-create model and train

model=LinearRegression()  #model creation
model.fit(area,price)     #train the model
```

```python
# step 4:-predict price based

predicted_price=model.predict([[3000]]) #give area
print(predicted_price)
```

```python
# step 5:-find the parameters of model slope and intercept

print(f"Slope (Coefficient): {model.coef_[0]}",predicted_price)
print(f"Intercept: {model.intercept_}")
```

```python

#step 6:-plot the data btw area and predicted price.

predicted_price=model.predict(area)
plt.scatter(area,price,color='blue',label='Actual Price')
plt.plot(area, predicted_price, color='red', label='Regression Line')
plt.xlabel('Area (square feet)')
plt.ylabel('Price (dollars)')
plt.title('Simple Linear Regression: Area vs. Price')
plt.legend()

plt.show()
```

### Example 2

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

engine_size = np.array([1000, 1500, 2000, 2500, 3000]).reshape(-1, 1)
car_price = np.array([20000, 25000, 30000, 35000, 40000])

model=LinearRegression()
model.fit(engine_size,car_price)

predicted_price=model.predict([[3000]])
print(predicted_price)

print(f"Slope (Coefficient): {model.coef_[0]}",predicted_price)
print(f"Intercept: {model.intercept_}")

predicted_price=model.predict(engine_size)
plt.scatter(engine_size,car_price,color='blue',label='Actual Price')
plt.plot(engine_size,predicted_price, color='red', label='Regression Line')
plt.xlabel('Engine_Size')
plt.ylabel('Car_Price')
plt.title('Simple Linear Regression: engine_size vs. Car_Price')

plt.legend()
plt.show()
```

# LOGISTIC REGRESSION

## Example 1

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# Generating a synthetic dataset
np.random.seed(0)
x=np.concatenate([np.random.normal(2,1,50),np.random.normal(6,1,50)]).reshape(-1,1)
y=np.array([0]*50+[1]*50)

# Train a logistic regression model
model=LogisticRegression()
model.fit(x,y)

# Plotting the data points
plt.figure(figsize=(10,6))
plt.scatter(x,y,color='red',label='Data Points')

# Plot the logistic regression curve
x_test=np.linspace(x.min(),x.max(),1000).reshape(-1,1)
y_prob=model.predict_proba(x_test)[:,1]
plt.plot(x_test,y_prob,color='blue',linewidth=2,label="Logistic Regressio Curve")

# Add labels and legend
plt.xlabel('Feature')
plt.ylabel('Probability')
plt.title("Logistic Regression")
plt.legend()
plt.grid()

#show the plot
plt.show()
```

## Example 2

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression,LinearRegression

# Generating a synthetic dataset
np.random.seed(0)
x=np.concatenate([np.random.normal(2,1,50),np.random.normal(6,1,50)]).reshape(-1,1)
y=np.array([0]*50+[1]*50)

# Train a logistic regression model
logistic_model=LogisticRegression()
logistic_model.fit(x,y)

# Train a linear regression model
linear_model=LinearRegression()
linear_model.fit(x,y)

# Plotting the data points
plt.figure(figsize=(10,6))
plt.scatter(x,y,color='red',label='data Points')

# Plot the logistic regression curve
x_test=np.linspace(x.min(),x.max(),1000).reshape(-1,1)
y_logistic_prob=logistic_model.predict_proba(x_test)[:,1]
plt.plot(x_test,y_logistic_prob,color='blue',linewidth=2,label="Logistic Regression Curve")

# Plot the linear regression line
y_linear_pred=linear_model.predict(x_test)
plt.plot(x_test,y_linear_pred,color='green',linestyle='--',linewidth=2,label='Linear Regression Line')

# Add labels and legend
plt.xlabel("Feature")
plt.ylabel("Output")
plt.title("Logisctic vs. Linear Regression")
plt.legend()

#show the plot
plt.show()
```
